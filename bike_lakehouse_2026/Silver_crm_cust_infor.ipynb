#INIT

```python
import pyspark.sql.functions as F
from pyspark.sql.functions import col, trim
from pyspark.sql.types import StringType
```

```python
RENAME_MAP = {
        "cst_id": "customer_id",
        "cst_key": "customer_key",
        "cst_firstname": "customer_firstname",
        "cst_lastname": "customer_lastname",
        "cst_marital_status": "marital_status",
        "cst_gndr": "gender",
        "cst_create_date": "create_date"
}
```

# Reading From Bronze Table 

```python
df = spark.table("workspace.bronze.crm_cust_info")
```

# Data Transformations

```python
#trim the string
#normalization for matial_status, gndr
#names are not friends

for field in df.schema.fields:
  if isinstance(field.dataType, StringType):
    df = df.withColumn(field.name, trim(col(field.name)))
```

## Normalization

```python
df = (
    df
    .withColumn(
        "cst_marital_status", 
        F.when(F.upper(F.col("cst_marital_status")) == "S", "Single")
         .when(F.upper(F.col("cst_marital_status")) == "M", "Married")
         .otherwise("n/a")
    )
    .withColumn(
        "cst_gndr", 
        F.when(F.upper(F.col("cst_gndr")) == "M", "Male")
         .when(F.upper(F.col("cst_gndr")) == "F", "Female")
         .otherwise("n/a")
    )
)
```

##Renaming The Columns

```python
for old_name, new_name in RENAME_MAP.items():
    df = df.withColumnRenamed(old_name, new_name)
```

```python
df.display()
```

# Write Into Silver Table

```python
(
    df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable("silver.crm_customer")
)
```
